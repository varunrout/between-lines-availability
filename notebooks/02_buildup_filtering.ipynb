# 02 – Build-Up Filtering

Apply the build-up filter and direction standardisation to the raw events.
Explore the resulting distribution of ball zones, possession phases and event types.


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt

from src.data.loader import load_competition
from src.features.buildup import filter_buildup
```

```python
events, frames, lineups, matches = load_competition(competition_id=55, season_id=43)
buildup_df = filter_buildup(events)
print(f'Build-up events: {len(buildup_df):,} from {len(events):,} total')
```

## Zone and phase distribution


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
buildup_df['ball_zone'].value_counts().plot(kind='bar', ax=axes[0], title='Ball Zone')
buildup_df['possession_phase'].value_counts().plot(kind='bar', ax=axes[1], title='Possession Phase')
plt.tight_layout()
plt.show()
```

## Ball position spatial distribution (standardised)


```python
try:
    from mplsoccer import Pitch
    pitch = Pitch(pitch_type='statsbomb', pitch_color='#1a1a2e', line_color='white')
    fig, ax = pitch.draw(figsize=(12, 8))
    ax.scatter(buildup_df['x'], buildup_df['y'], s=2, alpha=0.15, c='#00CC44')
    ax.set_title('Build-up Event Locations (standardised direction)', color='white')
    plt.show()
except ImportError:
    print('mplsoccer not installed – skipping pitch plot')
```

## Under-pressure events


```python
pressure_rate = buildup_df['under_pressure_flag'].mean() * 100
print(f'{pressure_rate:.1f}% of build-up events are under pressure')
print(buildup_df.groupby('possession_phase')['under_pressure_flag'].mean().mul(100).round(1))
```
